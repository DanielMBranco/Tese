```python
def parse_ts(df, src_col, dst_col="DATE"):
    c = f"`{src_col}`"  # protects column names with special chars

    return df.withColumn(
        dst_col,
        F.coalesce(
            F.expr(f"try_to_timestamp({c}, 'dd-MM-yyyy HH:mm:ss')"),  # 31-12-2023 16:00:00
            F.expr(f"try_to_timestamp({c}, 'dd/MM/yyyy HH:mm')"),     # 31/12/2023 16:00
            # optional extras, in case they appear in the same column:
            F.expr(f"try_to_timestamp({c}, 'dd/MM/yyyy HH:mm:ss')"),
            F.expr(f"try_to_timestamp({c}, 'dd-MM-yyyy HH:mm')")
        )
    )
```

```python
def cast_manual(c, d, e):
   return c.withColumn(d, expr(f"try_cast({d} as {e})"))
```
